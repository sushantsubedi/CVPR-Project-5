## KAT-ICL (Milestone) Report — ACT Transfer Cube

This notebook is an executable report of the current milestone implementation:

- Task: **ACT / ALOHA sim Transfer Cube** (`sim_transfer_cube_scripted`)

- Observation tokens: **top-view keypoints** from ViT-DINO patch descriptors (2D tokens)

- Decode + replay: decoded tokens are replayed in sim and success is computed from reward==4
